```python

# CAUSAL-X: THE BUTTERFLY EFFECT ENGINE
# Complete single-cell data science project

import sys
import subprocess
import importlib.util
import warnings
import os
import zipfile

warnings.filterwarnings("ignore")

# 1. Install missing dependencies
packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "seaborn": "seaborn",
    "networkx": "networkx",
    "sklearn": "scikit-learn",
    "plotly": "plotly",
    "IPython": "ipython"
}

missing = [
    package for module, package in packages.items()
    if importlib.util.find_spec(module) is None
]

if missing:
    subprocess.check_call(
        [sys.executable, "-m", "pip", "install", "-q", *missing]
    )

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx
import plotly.express as px

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.inspection import permutation_importance
from IPython.display import display

SEED = 42
N = 5000
rng = np.random.default_rng(SEED)
np.random.seed(SEED)

sns.set_theme(style="whitegrid")
OUTPUT_DIR = "causal_x_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("=" * 75)
print("CAUSAL-X: THE BUTTERFLY EFFECT ENGINE")
print("Supply-chain risk prediction and what-if impact simulation")
print("=" * 75)

# 2. Generate synthetic supply-chain data

supplier_reliability = np.clip(rng.normal(.93, .045, N), .65, 1)
demand_index = np.clip(rng.normal(100, 15, N), 50, 160)
weather_severity = rng.beta(2, 5, N)
transport_capacity = np.clip(rng.normal(.88, .08, N), .45, 1)
inventory_buffer = np.clip(rng.normal(.30, .10, N), .05, .65)

supplier_delay = np.maximum(
    0,
    2 + (1 - supplier_reliability) * 32
    + weather_severity * 7 + rng.normal(0, 2, N)
)

transport_delay = np.maximum(
    0,
    1 + weather_severity * 9
    + (1 - transport_capacity) * 16
    + supplier_delay * .24 + rng.normal(0, 1.5, N)
)

risk_logit = (
    -2.2 + (demand_index - 100) / 18
    + supplier_delay * .11 + transport_delay * .09
    - inventory_buffer * 3.2 + rng.normal(0, .25, N)
)

stockout_risk = 1 / (1 + np.exp(-risk_logit))

delivery_delay = np.maximum(
    0,
    supplier_delay * .48 + transport_delay * .72
    + stockout_risk * 8 + rng.normal(0, 1.8, N)
)

revenue = np.maximum(
    0,
    12000 + demand_index * 95
    - delivery_delay * 260 - stockout_risk * 4500
    + inventory_buffer * 1200 + rng.normal(0, 1300, N)
)

customer_satisfaction = np.clip(
    95 - delivery_delay * 1.25
    - stockout_risk * 22 + rng.normal(0, 3.5, N),
    0, 100
)

df = pd.DataFrame({
    "supplier_reliability": supplier_reliability,
    "demand_index": demand_index,
    "weather_severity": weather_severity,
    "transport_capacity": transport_capacity,
    "inventory_buffer": inventory_buffer,
    "supplier_delay": supplier_delay,
    "transport_delay": transport_delay,
    "stockout_risk": stockout_risk,
    "delivery_delay": delivery_delay,
    "revenue": revenue,
    "customer_satisfaction": customer_satisfaction
})

print("\n1. DATASET OVERVIEW")
print("Shape:", df.shape)
print("Missing values:", df.isna().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
display(df.head(10))
display(df.describe().T.round(3))

# 3. Exploratory data analysis

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

sns.histplot(df["delivery_delay"], kde=True, ax=axes[0, 0])
axes[0, 0].set_title("Delivery Delay Distribution")

sns.histplot(df["revenue"], kde=True, ax=axes[0, 1])
axes[0, 1].set_title("Revenue Distribution")

sns.scatterplot(
    data=df.sample(1200, random_state=SEED),
    x="delivery_delay", y="customer_satisfaction",
    hue="stockout_risk", palette="viridis",
    legend=False, ax=axes[1, 0]
)
axes[1, 0].set_title("Delivery Delay vs Satisfaction")

sns.scatterplot(
    data=df.sample(1200, random_state=21),
    x="supplier_delay", y="delivery_delay",
    hue="transport_delay", palette="magma",
    legend=False, ax=axes[1, 1]
)
axes[1, 1].set_title("Supplier Delay vs Delivery Delay")

plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt=".2f",
            cmap="coolwarm", center=0)
plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.show()

# 4. Build the assumed causal dependency graph

edges = [
    ("supplier_reliability", "supplier_delay"),
    ("weather_severity", "supplier_delay"),
    ("weather_severity", "transport_delay"),
    ("transport_capacity", "transport_delay"),
    ("supplier_delay", "transport_delay"),
    ("demand_index", "stockout_risk"),
    ("supplier_delay", "stockout_risk"),
    ("transport_delay", "stockout_risk"),
    ("inventory_buffer", "stockout_risk"),
    ("supplier_delay", "delivery_delay"),
    ("transport_delay", "delivery_delay"),
    ("stockout_risk", "delivery_delay"),
    ("delivery_delay", "revenue"),
    ("stockout_risk", "revenue"),
    ("demand_index", "revenue"),
    ("delivery_delay", "customer_satisfaction"),
    ("stockout_risk", "customer_satisfaction")
]

G = nx.DiGraph(edges)
pos = nx.spring_layout(G, seed=SEED, k=1.15)

upstream = {
    "supplier_reliability", "demand_index",
    "weather_severity", "transport_capacity", "inventory_buffer"
}
outcomes = {"revenue", "customer_satisfaction"}

node_colors = [
    "#90caf9" if n in upstream else
    "#a5d6a7" if n in outcomes else "#ffcc80"
    for n in G.nodes()
]

plt.figure(figsize=(15, 9))
nx.draw_networkx(
    G, pos, with_labels=True, node_size=2600,
    node_color=node_colors, font_size=8,
    font_weight="bold", arrowsize=18
)
plt.title("CAUSAL-X Dependency Graph")
plt.axis("off")
plt.tight_layout()
plt.show()

centrality = pd.DataFrame({
    "node": list(G.nodes()),
    "in_degree": [G.in_degree(n) for n in G.nodes()],
    "out_degree": [G.out_degree(n) for n in G.nodes()],
    "betweenness": [
        nx.betweenness_centrality(G)[n] for n in G.nodes()
    ]
}).sort_values("betweenness", ascending=False)

print("\n2. GRAPH CENTRALITY")
display(centrality)

# 5. Train and compare machine learning models

features = [
    "supplier_reliability", "demand_index", "weather_severity",
    "transport_capacity", "inventory_buffer", "supplier_delay",
    "transport_delay", "stockout_risk"
]

target = "delivery_delay"

X = df[features]
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=.20, random_state=SEED
)

models = {
    "Linear Regression": make_pipeline(
        StandardScaler(), LinearRegression()
    ),
    "Random Forest": RandomForestRegressor(
        n_estimators=160, max_depth=14,
        min_samples_leaf=3, random_state=SEED, n_jobs=-1
    ),
    "Hist Gradient Boosting": HistGradientBoostingRegressor(
        max_iter=180, learning_rate=.08,
        max_leaf_nodes=20, random_state=SEED
    )
}

results = []
fitted_models = {}
predictions = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_test)

    fitted_models[name] = model
    predictions[name] = pred

    results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_test, pred),
        "RMSE": np.sqrt(mean_squared_error(y_test, pred)),
        "R2": r2_score(y_test, pred)
    })

model_results = pd.DataFrame(results).sort_values("RMSE")
print("\n3. MODEL COMPARISON")
display(model_results.round(4))

best_name = model_results.iloc[0]["Model"]
best_model = fitted_models[best_name]
best_predictions = predictions[best_name]

print("Best model by test RMSE:", best_name)

plt.figure(figsize=(8, 6))
plt.scatter(y_test, best_predictions, alpha=.45)
limits = [
    min(y_test.min(), best_predictions.min()),
    max(y_test.max(), best_predictions.max())
]
plt.plot(limits, limits, "r--")
plt.xlabel("Actual Delivery Delay")
plt.ylabel("Predicted Delivery Delay")
plt.title(f"Actual vs Predicted: {best_name}")
plt.tight_layout()
plt.show()

residuals = y_test.to_numpy() - best_predictions

plt.figure(figsize=(8, 5))
sns.histplot(residuals, kde=True)
plt.title("Prediction Residual Distribution")
plt.xlabel("Actual minus predicted")
plt.tight_layout()
plt.show()

# 6. Predictive feature importance

perm = permutation_importance(
    best_model, X_test, y_test,
    n_repeats=5, random_state=SEED,
    scoring="neg_mean_absolute_error", n_jobs=-1
)

importance = pd.DataFrame({
    "Feature": features,
    "Importance": perm.importances_mean,
    "Standard Deviation": perm.importances_std
}).sort_values("Importance", ascending=False)

print("\n4. PERMUTATION FEATURE IMPORTANCE")
display(importance)

plt.figure(figsize=(9, 6))
sns.barplot(data=importance, x="Importance", y="Feature")
plt.title("Predictive Feature Importance")
plt.tight_layout()
plt.show()

# 7. Structural simulation function

def simulate_system(
    supplier_reliability=.93,
    demand_index=100,
    weather_severity=.28,
    transport_capacity=.88,
    inventory_buffer=.30,
    seed=123,
    samples=4000
):
    """Simulate the synthetic system under specified input conditions."""
    r = np.random.default_rng(seed)

    sr = np.full(samples, np.clip(supplier_reliability, .01, 1))
    demand = np.full(samples, np.clip(demand_index, 1, 200))
    weather = np.full(samples, np.clip(weather_severity, 0, 1))
    capacity = np.full(samples, np.clip(transport_capacity, .01, 1))
    inventory = np.full(samples, np.clip(inventory_buffer, 0, 1))

    sd = np.maximum(
        0, 2 + (1-sr)*32 + weather*7 + r.normal(0, 2, samples)
    )

    td = np.maximum(
        0, 1 + weather*9 + (1-capacity)*16
        + sd*.24 + r.normal(0, 1.5, samples)
    )

    risk_logit = (
        -2.2 + (demand-100)/18 + sd*.11
        + td*.09 - inventory*3.2
        + r.normal(0, .25, samples)
    )

    risk = 1 / (1 + np.exp(-risk_logit))

    delay = np.maximum(
        0, sd*.48 + td*.72 + risk*8
        + r.normal(0, 1.8, samples)
    )

    rev = np.maximum(
        0, 12000 + demand*95 - delay*260
        - risk*4500 + inventory*1200
        + r.normal(0, 1300, samples)
    )

    satisfaction = np.clip(
        95 - delay*1.25 - risk*22
        + r.normal(0, 3.5, samples), 0, 100
    )

    return pd.DataFrame({
        "supplier_delay": sd,
        "transport_delay": td,
        "stockout_risk": risk,
        "delivery_delay": delay,
        "revenue": rev,
        "customer_satisfaction": satisfaction
    })

# 8. What-if scenario analysis

base = {
    "supplier_reliability": .93,
    "demand_index": 100,
    "weather_severity": .28,
    "transport_capacity": .88,
    "inventory_buffer": .30
}

scenarios = {
    "Baseline": base,
    "Supplier reliability falls": {
        **base, "supplier_reliability": .85
    },
    "Severe weather": {
        **base, "weather_severity": .65
    },
    "Transport capacity falls": {
        **base, "transport_capacity": .78
    },
    "Demand increases": {
        **base, "demand_index": 115
    },
    "Inventory buffer improves": {
        **base, "inventory_buffer": .40
    },
    "Supplier reliability improves": {
        **base, "supplier_reliability": .97
    }
}

outcomes = [
    "supplier_delay", "transport_delay", "stockout_risk",
    "delivery_delay", "revenue", "customer_satisfaction"
]

scenario_samples = {}
scenario_rows = []

for name, parameters in scenarios.items():
    simulated = simulate_system(
        **parameters, seed=999, samples=6000
    )
    scenario_samples[name] = simulated
    scenario_rows.append({
        "Scenario": name,
        **simulated[outcomes].mean().to_dict()
    })

scenario_table = pd.DataFrame(scenario_rows).set_index("Scenario")
baseline_values = scenario_table.loc["Baseline"]

impact = scenario_table.subtract(baseline_values, axis="columns")
impact_pct = (
    impact.div(baseline_values.replace(0, np.nan), axis="columns") * 100
)

print("\n5. SCENARIO OUTCOMES")
display(scenario_table.round(3))

print("\n6. ABSOLUTE IMPACT VS BASELINE")
display(impact.round(3))

print("\n7. PERCENT CHANGE VS BASELINE")
display(impact_pct.round(2))

plt.figure(figsize=(13, 7))
sns.heatmap(impact, cmap="RdYlGn_r", center=0, annot=True, fmt=".2f")
plt.title("What-If Scenario Impact")
plt.tight_layout()
plt.show()

plt.figure(figsize=(13, 7))
sns.heatmap(impact_pct, cmap="RdYlGn_r", center=0, annot=True, fmt=".1f")
plt.title("Percentage Change from Baseline")
plt.tight_layout()
plt.show()

long_scenarios = scenario_table.reset_index().melt(
    id_vars="Scenario",
    value_vars=[
        "delivery_delay", "stockout_risk", "customer_satisfaction"
    ],
    var_name="Outcome",
    value_name="Mean Value"
)

fig = px.bar(
    long_scenarios, x="Scenario", y="Mean Value",
    color="Outcome", barmode="group",
    title="Interactive What-If Scenario Comparison"
)
fig.update_layout(template="plotly_white", xaxis_tickangle=-25)
fig.show()

# 9. Simulation uncertainty intervals

interval_rows = []

for name, simulated in scenario_samples.items():
    row = {"Scenario": name}

    for outcome in [
        "delivery_delay", "revenue", "customer_satisfaction"
    ]:
        row[outcome + "_q05"] = simulated[outcome].quantile(.05)
        row[outcome + "_median"] = simulated[outcome].median()
        row[outcome + "_q95"] = simulated[outcome].quantile(.95)

    interval_rows.append(row)

intervals = pd.DataFrame(interval_rows).set_index("Scenario")

print("\n8. SIMULATION PERCENTILE INTERVALS")
display(intervals.round(3))

# 10. Sensitivity analysis

sensitivity_inputs = {
    "supplier_reliability": (.93, .02),
    "demand_index": (100, 5),
    "weather_severity": (.28, .05),
    "transport_capacity": (.88, .05),
    "inventory_buffer": (.30, .05)
}

sensitivity_rows = []

for variable, (base_value, step) in sensitivity_inputs.items():
    plus_parameters = dict(base)
    minus_parameters = dict(base)

    plus_parameters[variable] = base_value + step
    minus_parameters[variable] = base_value - step

    plus = simulate_system(
        **plus_parameters, seed=777, samples=6000
    )
    minus = simulate_system(
        **minus_parameters, seed=777, samples=6000
    )

    for outcome in [
        "delivery_delay", "revenue", "customer_satisfaction"
    ]:
        effect = (
            plus[outcome].mean() - minus[outcome].mean()
        ) / (2 * step)

        sensitivity_rows.append({
            "Input": variable,
            "Outcome": outcome,
            "Sensitivity per unit": effect
        })

sensitivity = pd.DataFrame(sensitivity_rows)

print("\n9. SENSITIVITY ANALYSIS")
display(sensitivity.round(3))

sensitivity_pivot = sensitivity.pivot(
    index="Input", columns="Outcome",
    values="Sensitivity per unit"
)

plt.figure(figsize=(10, 6))
sns.heatmap(
    sensitivity_pivot, cmap="vlag",
    center=0, annot=True, fmt=".2f"
)
plt.title("Input Sensitivity Analysis")
plt.tight_layout()
plt.show()

# 11. Propagation paths

print("\n10. CAUSAL GRAPH PATHS")

for source in [
    "supplier_reliability", "weather_severity", "inventory_buffer"
]:
    print(f"\nStarting from: {source}")

    for target_node in [
        "delivery_delay", "revenue", "customer_satisfaction"
    ]:
        try:
            path = nx.shortest_path(G, source, target_node)
            print(" -> ".join(path))
        except nx.NetworkXNoPath:
            print(f"No path to {target_node}")

# 12. Automated project findings

print("\n" + "=" * 75)
print("CAUSAL-X INVESTIGATION SUMMARY")
print("=" * 75)

print("Number of observations:", len(df))
print("Graph nodes:", G.number_of_nodes())
print("Graph edges:", G.number_of_edges())
print("Best predictive model:", best_name)
print("MAE:", round(mean_absolute_error(y_test, best_predictions), 4))
print("RMSE:", round(np.sqrt(mean_squared_error(y_test, best_predictions)), 4))
print("R2:", round(r2_score(y_test, best_predictions), 4))

print("\nDelivery-delay change compared with baseline:")
for name in impact.index:
    print(f"{name}: {impact.loc[name, 'delivery_delay']:+.3f}")

print("\nInterpretation notes:")
print("- The dataset is synthetic.")
print("- The dependency graph is assumed by design.")
print("- Predictive importance and correlation are not causal proof.")
print("- Percentile intervals represent simulated outcome variability.")

# 13. Export datasets, results, and report

df.to_csv(f"{OUTPUT_DIR}/causal_x_dataset.csv", index=False)
model_results.to_csv(f"{OUTPUT_DIR}/model_comparison.csv", index=False)
importance.to_csv(f"{OUTPUT_DIR}/permutation_importance.csv", index=False)
scenario_table.to_csv(f"{OUTPUT_DIR}/scenario_outcomes.csv")
impact.to_csv(f"{OUTPUT_DIR}/scenario_absolute_impact.csv")
impact_pct.to_csv(f"{OUTPUT_DIR}/scenario_percent_impact.csv")
sensitivity.to_csv(f"{OUTPUT_DIR}/sensitivity_analysis.csv", index=False)
intervals.to_csv(f"{OUTPUT_DIR}/scenario_intervals.csv")

report_path = f"{OUTPUT_DIR}/investigation_report.txt"

with open(report_path, "w", encoding="utf-8") as report:
    report.write("CAUSAL-X Investigation Report\n")
    report.write("=" * 35 + "\n")
    report.write(f"Observations: {len(df)}\n")
    report.write(f"Best model: {best_name}\n")
    report.write(
        f"MAE: {mean_absolute_error(y_test, best_predictions):.4f}\n"
    )
    report.write(
        f"RMSE: {np.sqrt(mean_squared_error(y_test, best_predictions)):.4f}\n"
    )
    report.write(f"R2: {r2_score(y_test, best_predictions):.4f}\n\n")
    report.write("Scenario outcomes:\n")
    report.write(scenario_table.to_string())
    report.write("\n\nScenario impact:\n")
    report.write(impact.to_string())
    report.write("\n\nSensitivity analysis:\n")
    report.write(sensitivity.to_string())
    report.write(
        "\n\nLimitations: synthetic data and assumed structural equations.\n"
    )

zip_path = "causal_x_results.zip"

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as archive:
    for filename in os.listdir(OUTPUT_DIR):
        archive.write(
            os.path.join(OUTPUT_DIR, filename),
            arcname=filename
        )

print("\n11. EXPORTED FILES")
for filename in sorted(os.listdir(OUTPUT_DIR)):
    print(os.path.join(OUTPUT_DIR, filename))

print("\nResults archive:", zip_path)
print("\nCAUSAL-X EXECUTION FINISHED")
```
